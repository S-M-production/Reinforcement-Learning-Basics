```python
import numpy as np
```

COPY

```python
#When using copy, assignment doesnt assign pointer, rather it assigns a new object
arr = np.array([1, 2, 3, 4, 5])
x = arr.copy()
arr[0] = 42

print(arr)
print(x)
```

view

```python
 #View creates a reference, or pointer to original object
arr = np.array([1, 2, 3, 4, 5])
x = arr.view()
arr[0] = 42

print(arr)
print(x)
```

Check if array owns its own object

```python
#If it returns none, that means it owns its own copy, if its [] or returns a array, then its a view or reference
arr = np.array([1, 2, 3, 4, 5])

x = arr.copy()
y = arr.view()

print(x.base)
print(y.base)
```
